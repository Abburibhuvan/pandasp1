### **6. Get the list of countries where the 'Profit' and 'Shipping Cost's are greater than or equal to 2000 and 300 respectively.**

```python
df["Profit"]  = df["Profit"].astype(str).str.strip()
```

```python
if df["Profit"].astype(str).str.contains("$",regex = False).any():
    df["Profit"] = df["Profit"].astype(str).str.replace("$","",regex = False).copy()
    
if df["Profit"].astype(str).str.contains(",",regex = False).any():
    df["Profit"] = df["Profit"].astype(str).str.replace(",","",regex = False).copy()

if df["Profit"].astype(str).str.contains("(",regex = False).any():
    df["Profit"] = df["Profit"].astype(str).str.replace("(","",regex = False).copy()

if df["Profit"].astype(str).str.contains(")",regex = False).any():
    df["Profit"] = df["Profit"].astype(str).str.replace(")","",regex = False).copy()
```

```python
df["Profit"] = pd.to_numeric(df["Profit"], errors="coerce")
```

```python
if df["Shipping Cost"].astype(str).str.contains("$", regex=False).any():
    df["Shipping Cost"] = df["Shipping Cost"].astype(str).str.replce("$",regex = False)
```

```python
df["Shipping Cost"] = pd.to_numeric(df["Shipping Cost"], errors = "coerce")
```

### **8. Find the overall rank of "India" where the 'Profit' is maximum under the category 'Technology'.**

```python
df["Profit"]  = df["Profit"].astype(str).str.strip()
```

```python
if df["Profit"].astype(str).str.contains("$",regex = False).any():
    df["Profit"] = df["Profit"].astype(str).str.replace("$","",regex = False).copy()
    
if df["Profit"].astype(str).str.contains(",",regex = False).any():
    df["Profit"] = df["Profit"].astype(str).str.replace(",","",regex = False).copy()

if df["Profit"].astype(str).str.contains("(",regex = False).any():
    df["Profit"] = df["Profit"].astype(str).str.replace("(","",regex = False).copy()

if df["Profit"].astype(str).str.contains(")",regex = False).any():
    df["Profit"] = df["Profit"].astype(str).str.replace(")","",regex = False).copy()
```

```python
df["Profit"] = pd.to_numeric(df["Profit"], errors="coerce")
```

```python
tech = df[df["Category"] == "Technology"]

country_profit = tech.groupby("Country")["Profit"].sum().reset_index()

country_profit["Rank"] = country_profit["Profit"].rank(method="dense", ascending=False)

india_rank = country_profit[country_profit["Country"] == "India"]["Rank"].values[0]

print(india_rank)
```
